# Sprint 2 - Data Profiling and Data Quality

This notebook profiles, cleans, standardizes, and validates the six procurement datasets used in the project.

**Datasets:** `suppliers.csv`, `contracts.csv`, `purchase_orders.xlsx`, `inventory.json`, `warehouse.json`, `invoices.xml`.


```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("../../datasets")

suppliers = pd.read_csv(DATA_DIR / "suppliers.csv")
contracts = pd.read_csv(DATA_DIR / "contracts.csv")
purchase_orders = pd.read_excel(DATA_DIR / "purchase_orders.xlsx")
inventory = pd.read_json(DATA_DIR / "inventory.json")
warehouse = pd.read_json(DATA_DIR / "warehouse.json")
invoices = pd.read_xml(DATA_DIR / "invoices.xml")

print("All datasets loaded successfully!")
print("\nDataset sizes:")
print("Suppliers:", suppliers.shape)
print("Contracts:", contracts.shape)
print("Purchase Orders:", purchase_orders.shape)
print("Inventory:", inventory.shape)
print("Warehouse:", warehouse.shape)
print("Invoices:", invoices.shape)
```

## 1. Dataset Overview

```python
datasets = {
    "Purchase Orders": purchase_orders,
    "Suppliers": suppliers,
    "Warehouses": warehouse,
    "Contracts": contracts,
    "Inventory": inventory,
    "Invoices": invoices
}

for name, df in datasets.items():
    print(f"\n{name}")
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("-" * 40)
```

## 2. Column Names and Data Types

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.dtypes)
```

## 3. Missing-Value Analysis

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```

## 4. Duplicate Row Analysis

```python
for name, df in datasets.items():
    print(f"{name} - duplicate rows:", df.duplicated().sum())
    print("-" * 40)
```

## 5. Primary-Key Validation

`PO_ID` is the primary key for Purchase Orders. `Product_ID` is not the primary key because the same product can occur in multiple purchase orders.


```python
primary_keys = {
    "Purchase Orders": "PO_ID",
    "Suppliers": "Supplier_ID",
    "Warehouses": "Warehouse_ID",
    "Contracts": "Contract_ID",
    "Inventory": "Inventory_ID",
    "Invoices": "Invoice_ID"
}

for name, df in datasets.items():
    key = primary_keys[name]
    print(name)
    print("Primary Key:", key)
    print("Duplicate", key, ":", df[key].duplicated().sum())
    print()
```

## 6. Basic Statistics

```python
for name, df in datasets.items():
    print(f"\n{name}")
    display(df.describe())
```

## 7. Overall Profiling Report

```python
profiling_report = []

for name, df in datasets.items():
    for column in df.columns:
        profiling_report.append({
            "Dataset": name,
            "Column": column,
            "Data_Type": str(df[column].dtype),
            "Total_Records": len(df),
            "Missing_Values": int(df[column].isnull().sum()),
            "Missing_%": round(df[column].isnull().mean() * 100, 2),
            "Unique_Values": int(df[column].nunique()),
            "Duplicate_Values": int(df[column].duplicated().sum())
        })

profiling_report = pd.DataFrame(profiling_report)
display(profiling_report)
```

## 8. Missing-Value Report

```python
missing_report = []

for name, df in datasets.items():
    for column in df.columns:
        missing = int(df[column].isnull().sum())
        missing_report.append({
            "Dataset": name,
            "Column": column,
            "Missing_Values": missing,
            "Missing_Percentage": round((missing / len(df)) * 100, 2)
        })

missing_report = pd.DataFrame(missing_report)
display(missing_report)
```

## 9. Duplicate-Key Report

```python
duplicate_report = []

for name, df in datasets.items():
    key = primary_keys[name]
    duplicate_report.append({
        "Dataset": name,
        "Primary_Key": key,
        "Total_Rows": len(df),
        "Duplicate_Rows": int(df.duplicated().sum()),
        "Duplicate_Keys": int(df[key].duplicated().sum())
    })

duplicate_report = pd.DataFrame(duplicate_report)
display(duplicate_report)
```

## 10. Referential-Integrity Checks

```python
def check_foreign_key(child_df, child_column, parent_df, parent_column, ignore_null=True):
    child_values = child_df[child_column]
    if ignore_null:
        child_values = child_values.dropna()
    invalid = child_values[~child_values.isin(parent_df[parent_column])]
    return int(len(invalid))

print("Purchase Orders → Suppliers:",
      check_foreign_key(purchase_orders, "Supplier_ID", suppliers, "Supplier_ID"))

print("Purchase Orders → Warehouses:",
      check_foreign_key(purchase_orders, "Warehouse_ID", warehouse, "Warehouse_ID"))

print("Inventory → Purchase Orders:",
      check_foreign_key(inventory, "PO_ID", purchase_orders, "PO_ID"))

print("Inventory → Warehouses:",
      check_foreign_key(inventory, "Warehouse_ID", warehouse, "Warehouse_ID"))

print("Invoices → Purchase Orders:",
      check_foreign_key(invoices, "PO_ID", purchase_orders, "PO_ID"))

print("Invoices → Suppliers:",
      check_foreign_key(invoices, "Supplier_ID", suppliers, "Supplier_ID"))
```

## 11. Create Clean Copies

```python
suppliers_clean = suppliers.copy()
contracts_clean = contracts.copy()
warehouses_clean = warehouse.copy()
purchase_orders_clean = purchase_orders.copy()
inventory_clean = inventory.copy()
invoices_clean = invoices.copy()
```

## 12. Standardize Column Names

```python
def standardize_columns(df):
    df = df.copy()
    df.columns = (
        df.columns
        .str.strip()
        .str.replace(" ", "_", regex=False)
        .str.replace("-", "_", regex=False)
    )
    return df

suppliers_clean = standardize_columns(suppliers_clean)
warehouses_clean = standardize_columns(warehouses_clean)
purchase_orders_clean = standardize_columns(purchase_orders_clean)
contracts_clean = standardize_columns(contracts_clean)
inventory_clean = standardize_columns(inventory_clean)
invoices_clean = standardize_columns(invoices_clean)

suppliers_clean = suppliers_clean.drop_duplicates().copy()
warehouses_clean = warehouses_clean.drop_duplicates().copy()
purchase_orders_clean = purchase_orders_clean.drop_duplicates().copy()
contracts_clean = contracts_clean.drop_duplicates().copy()
inventory_clean = inventory_clean.drop_duplicates().copy()
invoices_clean = invoices_clean.drop_duplicates().copy()
```

## 13. Remove Extra Spaces from Text Fields

```python
def clean_text_columns(df):
    df = df.copy()
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].str.strip()
    return df

suppliers_clean = clean_text_columns(suppliers_clean)
warehouses_clean = clean_text_columns(warehouses_clean)
purchase_orders_clean = clean_text_columns(purchase_orders_clean)
inventory_clean = clean_text_columns(inventory_clean)
contracts_clean = clean_text_columns(contracts_clean)
invoices_clean = clean_text_columns(invoices_clean)
```

## 14. Handle Missing Values

- Missing `Contract_ID` is represented as `No Contract`.
- Missing `Paid_Date` is kept as missing because an unpaid invoice does not have a payment date. `Payment_Status` already records the payment state.


```python
purchase_orders_clean["Contract_ID"] = (
    purchase_orders_clean["Contract_ID"].fillna("No Contract")
)

print("Missing Contract_ID:",
      purchase_orders_clean["Contract_ID"].isnull().sum())

print("Missing Paid_Date before date conversion:",
      invoices_clean["Paid_Date"].isnull().sum())
```

## 15. Convert Date Columns

```python
date_columns = {
    "Purchase Orders": (purchase_orders_clean,
                        ["PO_Date", "Expected_Delivery_Date", "Actual_Receipt_Date"]),
    "Contracts": (contracts_clean, ["Start_Date", "End_Date"]),
    "Inventory": (inventory_clean, ["Snapshot_Date"]),
    "Invoices": (invoices_clean,
                 ["Invoice_Date", "Receipt_Date", "Due_Date", "Paid_Date"])
}

date_conversion_report = []

for dataset_name, (df, columns) in date_columns.items():
    for column in columns:
        original_missing = int(df[column].isnull().sum())
        df[column] = pd.to_datetime(df[column], errors="coerce")
        final_missing = int(df[column].isnull().sum())

        date_conversion_report.append({
            "Dataset": dataset_name,
            "Column": column,
            "Original_Missing": original_missing,
            "Final_Missing": final_missing,
            "New_Invalid_or_Unparseable": max(final_missing - original_missing, 0)
        })

date_conversion_report = pd.DataFrame(date_conversion_report)
display(date_conversion_report)
```

## 16. Check Data Types After Cleaning

```python
cleaned_datasets = {
    "Purchase Orders": purchase_orders_clean,
    "Suppliers": suppliers_clean,
    "Warehouses": warehouses_clean,
    "Contracts": contracts_clean,
    "Inventory": inventory_clean,
    "Invoices": invoices_clean
}

for name, df in cleaned_datasets.items():
    print(f"\n{name}")
    print(df.dtypes)
```

## 17. Check Remaining Missing Values

```python
for name, df in cleaned_datasets.items():
    remaining = df.isnull().sum()
    remaining = remaining[remaining > 0]
    print(f"\n{name}")
    print("No missing values." if remaining.empty else remaining)
```

## 18. Referential Integrity After Cleaning

```python
print("Purchase Orders → Suppliers:",
      check_foreign_key(purchase_orders_clean, "Supplier_ID",
                        suppliers_clean, "Supplier_ID"))

print("Purchase Orders → Warehouses:",
      check_foreign_key(purchase_orders_clean, "Warehouse_ID",
                        warehouses_clean, "Warehouse_ID"))

po_with_contract = purchase_orders_clean[
    purchase_orders_clean["Contract_ID"] != "No Contract"
]

print("Purchase Orders → Contracts:",
      check_foreign_key(po_with_contract, "Contract_ID",
                        contracts_clean, "Contract_ID"))

print("Inventory → Purchase Orders:",
      check_foreign_key(inventory_clean, "PO_ID",
                        purchase_orders_clean, "PO_ID"))

print("Inventory → Warehouses:",
      check_foreign_key(inventory_clean, "Warehouse_ID",
                        warehouses_clean, "Warehouse_ID"))

print("Invoices → Purchase Orders:",
      check_foreign_key(invoices_clean, "PO_ID",
                        purchase_orders_clean, "PO_ID"))

print("Invoices → Suppliers:",
      check_foreign_key(invoices_clean, "Supplier_ID",
                        suppliers_clean, "Supplier_ID"))
```

## 19. Purchase Order Range Validation

```python
po_range_checks = {
    "Negative Quantity": purchase_orders_clean["Quantity"] < 0,
    "Negative Unit Cost": purchase_orders_clean["Unit_Cost_INR"] < 0,
    "Invalid Import Duty %": (
        (purchase_orders_clean["Import_Duty_Percent"] < 0) |
        (purchase_orders_clean["Import_Duty_Percent"] > 100)
    ),
    "Negative Total PO Value": purchase_orders_clean["Total_PO_Value_INR"] < 0
}

for check_name, condition in po_range_checks.items():
    print(f"{check_name}: {int(condition.sum())}")
```

## 20. Inventory Range Validation

```python
inventory_range_checks = {
    "Negative Opening Stock": inventory_clean["Opening_Stock"] < 0,
    "Negative Stock In": inventory_clean["Stock_In"] < 0,
    "Negative Stock Out": inventory_clean["Stock_Out"] < 0,
    "Negative Closing Stock": inventory_clean["Closing_Stock"] < 0,
    "Invalid Capacity Utilization": (
        (inventory_clean["Capacity_Utilization_Percent"] < 0) |
        (inventory_clean["Capacity_Utilization_Percent"] > 100)
    )
}

for check_name, condition in inventory_range_checks.items():
    print(f"{check_name}: {int(condition.sum())}")
```

## 21. Invoice Range Validation

```python
invoice_range_checks = {
    "Negative Invoice Amount": invoices_clean["Invoice_Amount_INR"] < 0,
    "Negative Exchange Rate": invoices_clean["Exchange_Rate_to_INR"] < 0
}

for check_name, condition in invoice_range_checks.items():
    print(f"{check_name}: {int(condition.sum())}")
```

## 22. Inventory Consistency Validation

Business rule:

`Closing Stock = Opening Stock + Stock In - Stock Out`

The quality flag is calculated from the actual comparison instead of marking every row as valid automatically.


```python
inventory_clean["Calculated_Closing_Stock"] = (
    inventory_clean["Opening_Stock"]
    + inventory_clean["Stock_In"]
    - inventory_clean["Stock_Out"]
)

inventory_errors = inventory_clean[
    inventory_clean["Closing_Stock"] != inventory_clean["Calculated_Closing_Stock"]
].copy()

inventory_clean["Inventory_Quality_Flag"] = "Valid"
inventory_clean.loc[
    inventory_clean["Closing_Stock"] != inventory_clean["Calculated_Closing_Stock"],
    "Inventory_Quality_Flag"
] = "Invalid"

print("Inventory discrepancies:", len(inventory_errors))
print(inventory_clean["Inventory_Quality_Flag"].value_counts())
```

## 23. Purchase Order Calculation Validation

Business rule:

`Subtotal = Quantity × Unit Cost`

A tolerance of 1 INR is used for small rounding differences.


```python
purchase_orders_clean["Calculated_Subtotal"] = (
    purchase_orders_clean["Quantity"] *
    purchase_orders_clean["Unit_Cost_INR"]
)

po_errors = purchase_orders_clean[
    (
        purchase_orders_clean["Subtotal_INR"] -
        purchase_orders_clean["Calculated_Subtotal"]
    ).abs() > 1
].copy()

print("Purchase Order subtotal discrepancies:", len(po_errors))
```

## 24. Final Data Quality Report

```python
quality_checks = []

def add_quality_check(dataset, check, invalid_count):
    invalid_count = int(invalid_count)
    quality_checks.append({
        "Dataset": dataset,
        "Check": check,
        "Invalid_Count": invalid_count,
        "Status": "PASS" if invalid_count == 0 else "REVIEW"
    })

for name, df in cleaned_datasets.items():
    key = primary_keys[name]
    add_quality_check(name, f"Duplicate {key}", df[key].duplicated().sum())
    add_quality_check(name, "Remaining missing values", df.isnull().sum().sum())

add_quality_check(
    "Purchase Orders", "Supplier_ID foreign key",
    check_foreign_key(purchase_orders_clean, "Supplier_ID",
                      suppliers_clean, "Supplier_ID")
)
add_quality_check(
    "Purchase Orders", "Warehouse_ID foreign key",
    check_foreign_key(purchase_orders_clean, "Warehouse_ID",
                      warehouses_clean, "Warehouse_ID")
)
add_quality_check(
    "Purchase Orders", "Contract_ID foreign key excluding No Contract",
    check_foreign_key(po_with_contract, "Contract_ID",
                      contracts_clean, "Contract_ID")
)
add_quality_check(
    "Inventory", "PO_ID foreign key",
    check_foreign_key(inventory_clean, "PO_ID",
                      purchase_orders_clean, "PO_ID")
)
add_quality_check(
    "Inventory", "Warehouse_ID foreign key",
    check_foreign_key(inventory_clean, "Warehouse_ID",
                      warehouses_clean, "Warehouse_ID")
)
add_quality_check(
    "Invoices", "PO_ID foreign key",
    check_foreign_key(invoices_clean, "PO_ID",
                      purchase_orders_clean, "PO_ID")
)
add_quality_check(
    "Invoices", "Supplier_ID foreign key",
    check_foreign_key(invoices_clean, "Supplier_ID",
                      suppliers_clean, "Supplier_ID")
)

for check_name, condition in po_range_checks.items():
    add_quality_check("Purchase Orders", check_name, condition.sum())

for check_name, condition in inventory_range_checks.items():
    add_quality_check("Inventory", check_name, condition.sum())

for check_name, condition in invoice_range_checks.items():
    add_quality_check("Invoices", check_name, condition.sum())

add_quality_check("Inventory", "Closing stock calculation", len(inventory_errors))
add_quality_check("Purchase Orders", "Subtotal calculation", len(po_errors))

quality_report = pd.DataFrame(quality_checks)
display(quality_report)
```

## 25. Final Cleaned-Dataset Summary

```python
final_summary = []

for name, df in cleaned_datasets.items():
    final_summary.append({
        "Dataset": name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Duplicate_Rows": int(df.duplicated().sum()),
        "Missing_Values": int(df.isnull().sum().sum())
    })

final_summary = pd.DataFrame(final_summary)
display(final_summary)
```
