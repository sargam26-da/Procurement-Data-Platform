Product_ID is not the primary key of Purchase Orders. A product can appear in many purchase orders.

```python
#Checking Foreign key
def check_foreign_key(child_df, child_column, parent_df, parent_column):
    invalid = child_df[
        ~child_df[child_column].isin(parent_df[parent_column])
    ]

    return len(invalid)
```

```python
print("Purchase Orders → Suppliers:",
      check_foreign_key(
          purchase_orders,
          "Supplier_ID",
          suppliers,
          "Supplier_ID"
      ))

print("Purchase Orders → Warehouses:",
      check_foreign_key(
          purchase_orders,
          "Warehouse_ID",
          warehouses,
          "Warehouse_ID"
      ))

print("Inventory → Purchase Orders:",
      check_foreign_key(
          inventory,
          "PO_ID",
          purchase_orders,
          "PO_ID"
      ))

print("Inventory → Warehouses:",
      check_foreign_key(
          inventory,
          "Warehouse_ID",
          warehouses,
          "Warehouse_ID"
      ))

print("Invoices → Purchase Orders:",
      check_foreign_key(
          invoices,
          "PO_ID",
          purchase_orders,
          "PO_ID"
      ))

print("Invoices → Suppliers:",
      check_foreign_key(
          invoices,
          "Supplier_ID",
          suppliers,
          "Supplier_ID"
      ))
```

Clean and Standardize

```python
# Create copies so the original datasets are not changed
suppliers_clean = suppliers.copy()
contracts_clean = contracts.copy()
warehouses_clean = warehouses.copy()
purchase_orders_clean = purchase_orders.copy()
inventory_clean = inventory.copy()
invoices_clean = invoices.copy()
```

```python
#Standardize column names
def standardize_columns(df):
    df.columns = (
        df.columns
        .str.strip()
        .str.replace(" ", "_")
        .str.replace("-", "_")
    )
    return df

suppliers_clean = standardize_columns(suppliers_clean)
warehouses_clean = standardize_columns(warehouses_clean)
purchase_orders_clean = standardize_columns(purchase_orders_clean)
contracts_clean = contracts.drop_duplicates(contracts_clean)
inventory_clean = standardize_columns(inventory_clean)
invoices_clean = standardize_columns(invoices_clean)
```

```python
#Remove extra spaces from text fields
def clean_text_columns(df):
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].str.strip()
    return df

suppliers_clean = clean_text_columns(suppliers_clean)
warehouses_clean = clean_text_columns(warehouses_clean)
purchase_orders_clean = clean_text_columns(purchase_orders_clean)
inventory_clean = clean_text_columns(inventory_clean)
invoices_clean = clean_text_columns(invoices_clean)
```

Data Quality Validation

```python
#Referential Integrity
def check_foreign_key(child_df, child_column, parent_df, parent_column):
    invalid = child_df[
        ~child_df[child_column].isin(parent_df[parent_column])
    ]
    return len(invalid)
```

```python
print("Purchase Orders → Suppliers:",
      check_foreign_key(
          purchase_orders_clean,
          "Supplier_ID",
          suppliers_clean,
          "Supplier_ID"
      ))

print("Purchase Orders → Warehouses:",
      check_foreign_key(
          purchase_orders_clean,
          "Warehouse_ID",
          warehouses_clean,
          "Warehouse_ID"
      ))

print("Purchase Orders → Contracts:",
      check_foreign_key(
          purchase_orders_clean,
          "Contract_ID",
          contracts_clean,
          "Contract_ID"
      ))

print("Inventory → Purchase Orders:",
      check_foreign_key(
          inventory_clean,
          "PO_ID",
          purchase_orders_clean,
          "PO_ID"
      ))

print("Inventory → Warehouses:",
      check_foreign_key(
          inventory_clean,
          "Warehouse_ID",
          warehouses_clean,
          "Warehouse_ID"
      ))

print("Invoices → Purchase Orders:",
      check_foreign_key(
          invoices_clean,
          "PO_ID",
          purchase_orders_clean,
          "PO_ID"
      ))

print("Invoices → Suppliers:",
      check_foreign_key(
          invoices_clean,
          "Supplier_ID",
          suppliers_clean,
          "Supplier_ID"
      ))
```

the Purchase Orders → Contracts check will count those as invalid.

That is expected, because "No Contract" is not an actual Contract_ID.

Therefore, for this particular check, exclude "No Contract":

```python
po_with_contract = purchase_orders_clean[
    purchase_orders_clean["Contract_ID"] != "No Contract"
]

print("Purchase Orders → Contracts:",
      check_foreign_key(
          po_with_contract,
          "Contract_ID",
          contracts_clean,
          "Contract_ID"
      ))
```

```python
#Range Validation
print("Negative Quantity:",
      (purchase_orders_clean["Quantity"] < 0).sum())

print("Negative Unit Cost:",
      (purchase_orders_clean["Unit_Cost_INR"] < 0).sum())

print("Invalid Import Duty %:",
      ((purchase_orders_clean["Import_Duty_Percent"] < 0) |
       (purchase_orders_clean["Import_Duty_Percent"] > 100)).sum())

print("Negative Total PO Value:",
      (purchase_orders_clean["Total_PO_Value_INR"] < 0).sum())
```

```python
#inventory
print("Negative Opening Stock:",
      (inventory_clean["Opening_Stock"] < 0).sum())

print("Negative Stock In:",
      (inventory_clean["Stock_In"] < 0).sum())

print("Negative Stock Out:",
      (inventory_clean["Stock_Out"] < 0).sum())

print("Negative Closing Stock:",
      (inventory_clean["Closing_Stock"] < 0).sum())

print("Invalid Capacity Utilization:",
      ((inventory_clean["Capacity_Utilization_Percent"] < 0) |
       (inventory_clean["Capacity_Utilization_Percent"] > 100)).sum())
```

```python
#invoices
print("Negative Invoice Amount:",
      (invoices_clean["Invoice_Amount_INR"] < 0).sum())

print("Negative Exchange Rate:",
      (invoices_clean["Exchange_Rate_to_INR"] < 0).sum())
```

```python
#Inventory Consistency
inventory_clean["Calculated_Closing_Stock"] = (
    inventory_clean["Opening_Stock"]
    + inventory_clean["Stock_In"]
    - inventory_clean["Stock_Out"]
).clip(lower=0)
inventory_errors = inventory_clean[
    inventory_clean["Closing_Stock"] !=
    inventory_clean["Calculated_Closing_Stock"]
].copy()

print("Inventory discrepancies:", len(inventory_errors))
inventory_clean["Inventory_Quality_Flag"] = "Valid"
print(inventory_clean["Inventory_Quality_Flag"].value_counts())
```

```python
#Purchase Order Calculation
#Subtotal = Quantity × Unit Cost
purchase_orders_clean["Calculated_Subtotal"] = (
    purchase_orders_clean["Quantity"] *
    purchase_orders_clean["Unit_Cost_INR"]
)

po_errors = purchase_orders_clean[
    abs(
        purchase_orders_clean["Subtotal_INR"]
        - purchase_orders_clean["Calculated_Subtotal"]
    ) > 1
]

print("Purchase Order subtotal discrepancies:",
      len(po_errors))
```
